```python
print("Split: by_polishing")

for column in df_polishing.columns:
    empty_count = df_polishing[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```

```python
print("Split: from_title")

for column in df_title.columns:
    empty_count = df_title[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```

```python
print("Split: from_title_and_content")

for column in df_title_content.columns:
    empty_count = df_title_content[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```
