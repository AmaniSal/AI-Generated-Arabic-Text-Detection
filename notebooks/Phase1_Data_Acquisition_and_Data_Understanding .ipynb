Check each column in by_polishing, from_title, from_title_and_content for empty strings or text containing only whitespace.

```python
print("Split: by_polishing")

for column in df_polishing.columns:
    empty_count = df_polishing[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```

```python
print("Split: from_title")

for column in df_title.columns:
    empty_count = df_title[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```

```python
print("Split: from_title_and_content")

for column in df_title_content.columns:
    empty_count = df_title_content[column].astype(str).str.strip().eq("").sum()
    print(column, ":", empty_count)
```

```python
#Count the human-written texts and the AI-generated texts in the by_polishing split.
human_count = len(df_polishing) #Each row contains one original human-written abstract.
print("Human:", human_count)
ai_count = len(df_polishing)*4 #Each row contains four AI-generated versions,number of rows is multiplied by 4.
print ("AI:", ai_count)
```

```python
human_count = len(df_title) #Count the human-written texts and the AI-generated texts in the from_title split.
print("Human:", human_count)
ai_count = len(df_title)*4
print ("AI:", ai_count)
```

```python
human_count = len(df_title_content) #Count the human-written texts and the AI-generated texts in the from_title_and_content split.
print("Human:", human_count)
ai_count = len(df_title_content)*4
print ("AI:", ai_count)
```

```python
total_human = 2851 + 2963 + 2574 #Calculate the total number of human-written texts across all three splits.
total_ai = 11404 + 11852 + 10296 #Calculate the total number of AI-generated texts across all three splits.
total_texts = total_human + total_ai #Calculate the total number of text instances.

#Display the total counts.
print("Total Human:", total_human)
print("Total AI:", total_ai)
print("Total Texts:", total_texts)

#Calculate and display the percentage of Human and AI-generated texts.
print("Human Percentage:", (total_human / total_texts) * 100, "%")
print("AI Percentage:", (total_ai / total_texts) * 100, "%")
```
